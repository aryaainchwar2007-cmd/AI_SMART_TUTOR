```python
import pandas as pd
import numpy as np

# Load student performance dataset
df = pd.read_csv('../data/student_performance.csv')
print("Shape:", df.shape)
df.head()
```

```python
# Check missing values and duplicate rows
print("Missing values per feature:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Parametric descriptive statistics
df.describe().T[['mean', 'std', 'min', '25%', '50%', '75%', 'max']]
```

```python
# Detect outliers using IQR rule on study_hours
q1 = df['study_hours'].quantile(0.25)
q3 = df['study_hours'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[(df['study_hours'] < lower_bound) | (df['study_hours'] > upper_bound)]
print(f"IQR: {iqr:.2f}, Lower Bound: {lower_bound:.2f}, Upper Bound: {upper_bound:.2f}")
print(f"Detected {len(outliers)} outliers in study_hours.")
```
