### Objectives:
1. Configure clean academic figure aesthetics
2. Plot distributions (Histograms with KDE, Boxplots)
3. Analyze bivariate linear and non-linear relationships with Scatterplots
4. Compute and visualize Pearson correlation matrices via Heatmaps

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set uniform academic theme
sns.set_theme(style='whitegrid', palette='deep')
df = pd.read_csv('../data/iris.csv')
df.head()
```

```python
# 1. Histogram with Kernel Density Estimation (KDE)
fig, ax = plt.subplots(figsize=(8, 4.5))
sns.histplot(df['sepal_length'], kde=True, bins=15, color='#2563eb', ax=ax)
ax.set_title('Sepal Length Distribution (Gaussian & Modality Analysis)', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# 2. Bivariate Scatterplot with Category Hue
fig, ax = plt.subplots(figsize=(8, 4.5))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species', s=70, ax=ax)
ax.set_title('Petal Length vs Petal Width by Species', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# 3. Pearson Correlation Heatmap
fig, ax = plt.subplots(figsize=(7, 5))
corr = df.select_dtypes(include=['number']).corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, square=True, ax=ax)
ax.set_title('Pearson Correlation Heatmap (Linear Association)', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```
